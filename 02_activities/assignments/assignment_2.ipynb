**Description**  
Given the `root` of a binary tree containing non-negative values, check whether it contains a duplicate value. If a duplicate exists, return the duplicate value. If there are multiple duplicates, return the one whose duplicate is closest to the root. If no duplicate exists, return -1.

**Examples**
* Example 1:

    Output: 2

* Example 2:

    ![](./images/q1_ex2.png)

    Input: `root = [1, 10, 2, 3, 10, 12, 12]`

    Output: 10

* Example 3:

    ![](./images/q1_ex3.png)

    Input: `root = [10, 9, 8, 7]`

    Output: -1

**Problem 1 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
# Definition for a binary tree node.
# class TreeNode(object):
#     def __init__(self, val = 0, left = None, right = None):
#         self.val = val
#         self.left = left
#         self.right = right

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def is_duplicate(root: TreeNode) -> int:
    if root is None:
        return -1

    seen = set()
    queue = deque([root])

    while queue:
        node = queue.popleft()

        if node.val in seen:
            return node.val

        seen.add(node.val)

        if node.left is not None:
            queue.append(node.left)

        if node.right is not None:
            queue.append(node.right)

    return -1
```

```python
root = TreeNode(1)
root.left = TreeNode(4)
root.right = TreeNode(5)
root.left.left = TreeNode(5)

print(is_duplicate(root))
```

```python

# Test 1: No duplicates
root1 = TreeNode(1)
root1.right = TreeNode(4)
root1.right.left = TreeNode(7)

print("Test 1:", is_duplicate(root1))

# Test 2: Empty tree
print("Test 2:", is_duplicate(None))

# Test 3: Multiple duplicates
root3 = TreeNode(1)
root3.left = TreeNode(10)
root3.right = TreeNode(2)
root3.left.left = TreeNode(3)
root3.left.right = TreeNode(10)
root3.right.left = TreeNode(12)
root3.right.right = TreeNode(12)

print("Test 3:", is_duplicate(root3))
```

3. Walkthrough one example that your partner made and explain it.

# Your answer here
For Molly's first example, `s = "{[()]}"`, the expected output is `True`.

The function starts with an empty stack. It reads the opening brackets `{`, `[`, and `(` and adds each one to the stack. When it reaches `)`, it checks the last opening bracket, which is `(`. They match, so it removes `(` from the stack. The next closing bracket, `]`, matches `[`, and the final `}` matches `{`. By the end, the stack is empty, so the function returns `True`.


4. Copy the solution your partner wrote. 


```python
# Your answer here

def is_valid_brackets(s: str) -> bool:
    stack = []
    bracket_map = {')': '(', '}': '{', ']': '['}

    for char in s:
        if char in bracket_map.values():
            stack.append(char)
        elif char in bracket_map.keys():
            if not stack or stack[-1] != bracket_map[char]:
                return False
            stack.pop()

    return len(stack) == 0
```

```python
print(is_valid_brackets("{[()]}"))
print(is_valid_brackets("([)]"))
print(is_valid_brackets("(("))
```
